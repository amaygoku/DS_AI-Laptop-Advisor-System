```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"D:\DS_project\DS_AI-Laptop-Advisor-System\data\processed\laptops_features.csv")

df["norm_weight"] = (
    1 - df["Weight (kg)"].rank(pct=True)
)

df["base_portability_score"] = df["norm_weight"].round(3)
df["is_light"] = df["Weight (kg)"] < 1.6
df["is_heavy"] = df["Weight (kg)"] > 2.3

battery_mask = df["Battery (Wh)"].notna()
df["norm_battery"] = np.nan

if battery_mask.any():
    df.loc[battery_mask, "norm_battery"] = (
        df.loc[battery_mask, "Battery (Wh)"]
          .rank(pct=True, method="average")
    )

df["battery_score"] = df["norm_battery"]
df["battery_available"] = df["battery_score"].notna().astype(int)

df["office_score"] = (
    df["base_performance_score"] * 0.45 +
    df["norm_weight"] * 0.35 +
    df["battery_score"].fillna(0.5) * 0.20
)

df["portability_score"] = (
    df["norm_weight"] * 0.45 +
    df["battery_score"].fillna(0.5) * 0.35 +
    (1 - df["norm_screen"]) * 0.20
)

df["general_score"] = (
    df["office_score"] * 0.45 +
    df["base_performance_score"] * 0.35 +
    df["portability_score"] * 0.20
)

df["is_business_ready"] = df["office_score"] >= 0.6
df["is_ultrabook"] = (
    (df["norm_weight"] >= 0.7) &
    (df["battery_score"].fillna(0.5) >= 0.6)
)
df.to_csv(r"D:\DS_project\DS_AI-Laptop-Advisor-System\data\processed\laptops_features2.csv", index=False)
```
