# Score Distribution Analysis

Comparing old vs new score data to recommend threshold adjustments.

```python
import pandas as pd
import numpy as np

# Load data
df_old = pd.read_csv(r'data\processed\laptops_features2 (1).csv')
df_new = pd.read_csv(r'data\processed\laptops_features2.csv')

print(f"OLD DATA shape: {df_old.shape}")
print(f"NEW DATA shape: {df_new.shape}")
```

```python
# Find score columns
score_cols = [c for c in df_new.columns if 'score' in c.lower()]
print("Score columns:", score_cols)
```

```python
# Compare distributions
for col in sorted(score_cols):
    if col in df_old.columns and col in df_new.columns:
        print(f"\n{'='*80}")
        print(f"{col}")
        print(f"{'='*80}")
        
        old_stats = df_old[col].describe()
        new_stats = df_new[col].describe()
        
        comparison = pd.DataFrame({
            'OLD': old_stats,
            'NEW': new_stats,
            'CHANGE': new_stats - old_stats
        })
        
        print(comparison)
        
        change_pct = ((new_stats['mean'] - old_stats['mean']) / old_stats['mean'] * 100) if old_stats['mean'] != 0 else 0
        print(f"\nMean change: {change_pct:+.2f}%")
```
