```python
# --- STEP 1: SETUP & LOAD DATA ---
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set the aesthetic style of the plots
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (14, 8) # Bigger default size

# === IMPORTANT: UPLOAD YOUR FILE FIRST ===
# Option A: If you uploaded 'merged_laptops_final.csv' directly to Colab
file_path = 'FinalData.csv'

# Option B: If it's in your Google Drive (Uncomment lines below)
# from google.colab import drive
# drive.mount('/content/drive')
# file_path = '/content/drive/My Drive/Your_Folder/merged_laptops_final.csv'

try:
    df = pd.read_csv(file_path)
    print("✅ Data Loaded Successfully!")
    print(f"Shape: {df.shape}")
except FileNotFoundError:
    print("❌ Error: File not found. Please upload 'merged_laptops_final.csv' to Colab files.")

# --- STEP 2: DATA CLEANING (Just in case) ---
# Ensure numeric columns are actually numeric
numeric_cols = ['Price (VND)', 'RAM (GB)', 'Storage (GB)', 'Screen Size (inch)', 'Refresh Rate (Hz)', 'Weight (kg)']
for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')
# Xem thông tin các cột và kiểu dữ liệu
df.info()

# Kiểm tra giá trị bị thiếu (Missing values)
print("\nSố lượng giá trị thiếu:")
print(df.isnull().sum())

# Chuyển đổi kiểu dữ liệu (nếu cần thiết, ví dụ Price, RAM phải là số)
# Code crawler của bạn đã làm tốt việc này, nhưng kiểm tra lại không thừa.
numeric_cols = ['Price (VND)', 'RAM (GB)', 'Storage (GB)', 'Screen Size (inch)', 'Refresh Rate (Hz)', 'Weight (kg)']
for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

# Lọc bỏ các dòng rác (nếu còn sót)
df_clean = df.dropna(subset=['Price (VND)', 'CPU manufacturer'])
print(f"\nDữ liệu sạch sau khi lọc: {len(df_clean)} dòng")



# --- STEP 3: GENERATE GRAPHS ---

# 1. Price Distribution (Histogram)
plt.figure(figsize=(12, 6))
sns.histplot(data=df_clean, x='Price (VND)', kde=True, bins=40, color='teal')
plt.title('Distribution of Laptop Prices', fontsize=16)
plt.xlabel('Price (VND)', fontsize=12)
plt.ylabel('Count', fontsize=12)
# Format X-axis to millions (e.g., 20M)
current_values = plt.gca().get_xticks()
plt.gca().set_xticklabels(['{:,.0f}M'.format(x/1000000) for x in current_values])
plt.show()

# 2. Market Share by Brand (Bar Chart)
plt.figure(figsize=(14, 7))
brand_counts = df_clean['Manufacturer'].value_counts()
sns.barplot(x=brand_counts.index, y=brand_counts.values, palette='viridis')
plt.title('Number of Laptops by Manufacturer', fontsize=16)
plt.xlabel('Brand', fontsize=12)
plt.ylabel('Number of Models', fontsize=12)
plt.xticks(rotation=45)
plt.show()

# 3. Price vs. RAM (Box Plot)
plt.figure(figsize=(12, 6))
sns.boxplot(data=df_clean, x='RAM (GB)', y='Price (VND)', palette='Blues')
plt.title('Price Distribution by RAM Size', fontsize=16)
plt.ylabel('Price (VND)', fontsize=12)
# Format Y-axis
plt.gca().set_yticklabels(['{:,.0f}M'.format(x/1000000) for x in plt.gca().get_yticks()])
plt.show()

# 4. Price vs. Storage (Box Plot)
plt.figure(figsize=(12, 6))
sns.boxplot(data=df_clean, x='Storage (GB)', y='Price (VND)', palette='Greens')
plt.title('Price Distribution by Storage Capacity', fontsize=16)
plt.ylabel('Price (VND)', fontsize=12)
plt.gca().set_yticklabels(['{:,.0f}M'.format(x/1000000) for x in plt.gca().get_yticks()])
plt.show()

# 5. CPU Manufacturer Comparison (Box Plot)
# Filter only major players
cpu_df = df_clean[df_clean['CPU manufacturer'].isin(['Intel', 'AMD', 'Apple'])]
plt.figure(figsize=(10, 6))
sns.boxplot(data=cpu_df, x='CPU manufacturer', y='Price (VND)', palette='Set2')
plt.title('Price Comparison: Intel vs. AMD vs. Apple', fontsize=16)
plt.gca().set_yticklabels(['{:,.0f}M'.format(x/1000000) for x in plt.gca().get_yticks()])
plt.show()

# 6. Correlation Heatmap (Multivariate Analysis)
plt.figure(figsize=(10, 8))
# Select only numeric columns for correlation
corr_matrix = df_clean[numeric_cols].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Heatmap of Laptop Features', fontsize=16)
plt.show()

# 7. Pair Plot (Scatterplot Matrix) - NEW!
# Visualizes pairwise relationships between numeric variables, colored by Manufacturer
# We select a subset of important numeric columns for clarity
pair_cols = ['Price (VND)', 'RAM (GB)', 'Storage (GB)', 'Weight (kg)']
# Filter top 5 brands to keep the legend clean
top_5_brands = df_clean['Manufacturer'].value_counts().head(5).index
df_pair = df_clean[df_clean['Manufacturer'].isin(top_5_brands)]

pair_plot = sns.pairplot(df_pair[pair_cols + ['Manufacturer']], hue='Manufacturer', palette='tab10', height=2.5)
pair_plot.fig.suptitle('Pairwise Relationships of Top 5 Laptop Brands', y=1.02, fontsize=16)
plt.show()

# 8. Violin Plot: Price vs GPU Manufacturer - NEW!
# Shows the density of price distribution for different GPUs
plt.figure(figsize=(12, 6))
sns.violinplot(data=df_clean, x='GPU manufacturer', y='Price (VND)', palette='muted')
plt.title('Price Density by GPU Manufacturer', fontsize=16)
plt.gca().set_yticklabels(['{:,.0f}M'.format(x/1000000) for x in plt.gca().get_yticks()])
plt.show()

# 9. Top 10 Most Common CPU Models - NEW!
# Creates a 'Full CPU Name' column if not existing, or uses 'CPU brand modifier' + 'generation'
# Assuming 'CPU brand modifier' holds the model (e.g., 'Core i5') and generation is separate.
# Adjust logic based on your exact column data if needed.
# Let's try combining Manufacturer + Brand Modifier + Generation for a label
df_clean['CPU_Full_Label'] = df_clean['CPU manufacturer'].astype(str) + " " + df_clean['CPU brand modifier'].astype(str)
top_cpus = df_clean['CPU_Full_Label'].value_counts().head(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=top_cpus.values, y=top_cpus.index, palette='rocket')
plt.title('Top 10 Most Common CPU Models', fontsize=16)
plt.xlabel('Count', fontsize=12)
plt.show()

# 10. Scatter Plot: Weight vs. Screen Size (Colored by Price) - NEW!
plt.figure(figsize=(10, 6))
scatter = sns.scatterplot(
    data=df_clean,
    x='Screen Size (inch)',
    y='Weight (kg)',
    hue='Price (VND)',
    size='Price (VND)',
    sizes=(20, 200),
    palette='viridis',
    alpha=0.7
)
plt.title('Laptop Weight vs. Screen Size (Color/Size = Price)', fontsize=16)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0)
plt.show()

# --- STEP 4: SUMMARY STATS ---
print("\n" + "="*30)
print("       QUICK SUMMARY")
print("="*30)
print(f"Total Products: {len(df_clean)}")
print(f"Avg Price: {df_clean['Price (VND)'].mean():,.0f} VND")
print(f"Median Price: {df_clean['Price (VND)'].median():,.0f} VND")
print(f"Top 3 Brands: {', '.join(brand_counts.head(3).index)}")
print("="*30)
```

```python

# --- STEP 3: GENERATE GRAPHS (Refined Selection) ---

# 1. Numerical Features Histograms
numeric_features = ['RAM (GB)', 'Storage (GB)', 'Screen Size (inch)', 'Refresh Rate (Hz)', 'Weight (kg)', 'Price (VND)']

# Create subplots for histograms
fig, axes = plt.subplots(2, 3, figsize=(18, 12))
fig.suptitle('Distribution of Numerical Features', fontsize=20, y=1.02)

# Flatten axes array for easy iteration
axes = axes.flatten()

for i, feature in enumerate(numeric_features):
    if feature in df_clean.columns:
        sns.histplot(data=df_clean, x=feature, kde=True, ax=axes[i], color='teal', bins=20)
        axes[i].set_title(f'Distribution of {feature}', fontsize=14)
        axes[i].set_xlabel(feature, fontsize=12)
        axes[i].set_ylabel('Count', fontsize=12)

        # Specific formatting for Price
        if feature == 'Price (VND)':
            current_values = axes[i].get_xticks()
            axes[i].set_xticklabels(['{:,.0f}M'.format(x/1000000) for x in current_values], rotation=45)

plt.tight_layout()
plt.show()

# 2. Categorical Features Boxplots vs Price
categorical_features = ['Manufacturer', 'CPU manufacturer', 'GPU manufacturer', 'RAM Type']

# Create subplots for boxplots
# Adjusting layout: 2 rows, 2 columns
fig, axes = plt.subplots(2, 2, figsize=(18, 14))
fig.suptitle('Price Distribution by Categorical Features', fontsize=20, y=1.02)

axes = axes.flatten()

for i, feature in enumerate(categorical_features):
    if feature in df_clean.columns:
        # Sort by median price for better readability
        order = df_clean.groupby(feature)['Price (VND)'].median().sort_values().index

        sns.boxplot(data=df_clean, x=feature, y='Price (VND)', ax=axes[i], order=order, palette='viridis')
        axes[i].set_title(f'Price vs {feature}', fontsize=14)
        axes[i].set_xlabel(feature, fontsize=12)
        axes[i].set_ylabel('Price (VND)', fontsize=12)

        # Rotate x-labels if there are many categories (like Manufacturer)
        if len(df_clean[feature].unique()) > 5:
            axes[i].tick_params(axis='x', rotation=45)

        # Format Y-axis to Millions
        current_values = axes[i].get_yticks()
        axes[i].set_yticklabels(['{:,.0f}M'.format(x/1000000) for x in current_values])

plt.tight_layout()
plt.show()

# --- STEP 4: SUMMARY STATS ---
print("\n" + "="*30)
print("       QUICK SUMMARY")
print("="*30)
print(f"Total Products: {len(df_clean)}")
print(f"Avg Price: {df_clean['Price (VND)'].mean():,.0f} VND")
print(f"Median Price: {df_clean['Price (VND)'].median():,.0f} VND")
print("="*30)
```
